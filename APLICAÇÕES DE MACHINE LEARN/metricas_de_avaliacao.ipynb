```python
import random
```

```python
y_true = [random.randint(0, 1) for _ in range(10)]
y_pred = [random.randint(0, 1) for _ in range(10)]
```

```python
def contador(array,num):
    cont=0
    for num_array in array:
        if array[num_array] == num:
            cont+=1            
    return cont
            
```

```python
def tabela_confusão(y_true, y_pred):
    TP = FP = TN = FN = 0
    
    for real, pred in zip(y_true, y_pred):
        if real == 1 and pred == 1:
            TP += 1
        elif real == 0 and pred == 1:
            FP += 1
        elif real == 0 and pred == 0:
            TN += 1
        elif real == 1 and pred == 0:
            FN += 1
    return [TP,FP,TN,FN]



TP,FP,TN,FN = tabela_confusão(y_true, y_pred)
```

```python
print(f"True Positivo: {TP}, Falso Positivo: {FP}, True Negativo: {TN}, Falso Negativo: {FN}")

precision = TP / (TP + FP) if (TP + FP) > 0 else 0
print("Precision:", precision)

recall = TP / (TP + FN) if (TP + FN) > 0 else 0
print("Recall:", recall)

f1 = (2 * precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
print("F1:", f1)

especificidade = TN / (TN + FP) if (TN + FP) > 0 else 0
print("Especificidade:", especificidade)

fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
print("Taxa de Falsos Positivos:", fpr)

fnr = FN / (FN + TP) if (FN + TP) > 0 else 0
print("Taxa de Falsos Negativos:", fnr)

accuracy = (TP + TN) / (TP + TN + FP + FN)
print("Accuracy:", accuracy)
```
