    sem intervalos de classe

```python
import pandas as pd
import os
from collections import Counter
```

```python
dados_bruto = [
    2, 3, 4, 4, 5, 3, 4,
    5, 6, 5, 3, 1, 5, 5,
    1, 3, 4, 5, 5, 5, 3,
    2, 2, 5, 4, 4, 2, 3,
    5, 4, 5, 4, 2, 4, 9,
]   

```

```python
len(dados_bruto)
```

```python
contagem = Counter(dados_bruto)
print(dict(contagem))

```

```python
df_dados = pd.DataFrame(contagem.items(), columns=["Número", "Frequência"])
df_dados
```

```python
df_dados["Relativa"] = (df_dados["Frequência"] / len(dados_bruto)) * 100

df_dados["Relativa"] = df_dados["Relativa"].round(2)

print(df_dados)
```

```python
df_dados["Frequência Acumulada"] = df_dados["Frequência"].cumsum()
df_dados
```

<p>
a) Tendo em vista essa pesquisa, construa uma distribuição de frequência 
sem intervalos de classe


</p>

```python
df_dados["Acumulada Relativa"] = (df_dados["Frequência Acumulada"] / len(dados_bruto)) * 100

df_dados["Acumulada Relativa"] = df_dados["Acumulada Relativa"].round(2)

df_dados
```

---------------------------------------------------------

```python

valor_maximo = max(dados_bruto)
print(valor_maximo)

valor_minimo = min(dados_bruto)
print(valor_minimo)

amplitude_total = valor_maximo - valor_minimo
amplitude_total
```

```python
len(dados_bruto)
```

```python
import math
k = (1 + 3.3 * math.log10(len(dados_bruto)))
numero_de_classes = round(k)
numero_de_classes
```

```python
amplitude_total
```

```python
amplitude_do_intervalo = round( amplitude_total / numero_de_classes)
amplitude_do_intervalo
```

```python

bins = [valor_minimo + i*amplitude_do_intervalo for i in range(numero_de_classes+1)]
bins
```

```python
classes = pd.cut(dados_bruto, bins=bins, right=True, include_lowest=True)
classes
```

```python
frequencia_intervalos = classes.value_counts().sort_index()
frequencia_intervalos
```

```python

# Montar DataFrame
df_frequencia_intervalo = pd.DataFrame({
    "Intervalo": frequencia_intervalos.index.astype(str),
    "Frequência": frequencia_intervalos.values
})

```

```python
df_frequencia_intervalo["Relativa"] = (df_frequencia_intervalo["Frequência"] / len(dados_bruto) * 100).round(2)
```

```python
df_frequencia_intervalo["Acumulada"] = df_frequencia_intervalo["Frequência"].cumsum()
```

```python
df_frequencia_intervalo["Acumulada Relativa"] = df_frequencia_intervalo["Relativa"].cumsum().round(2)
```

```python
df_frequencia_intervalo
#deu quebrado
```

# Exercício 3

```python

dados = {
    "Tempo (h)": ["0-4", "4-8", "8-12", "12-16", "16-20"],
    "Nº de pacientes": [8, 15, 24, 20, 13]
}

# Criando DataFrame
df_pacientes = pd.DataFrame(dados)

df_pacientes
```

```python
df_pacientes["Relativa"] = (df_pacientes["Nº de pacientes"] / len(dados) * 100).round(2)
```

```python
df_pacientes["Acumulada"] = df_pacientes["Nº de pacientes"].cumsum()
```

```python
df_pacientes["Acumulada Relativa"] = df_pacientes["Relativa"].cumsum().round(2)
df_pacientes
```

```python
limites = [list(map(int, intervalos.split("-"))) for intervalos in df_pacientes["Tempo (h)"]]

limite_inferior = [l[0] for l in limites]
print(limite_inferior)
limite_superior = [l[1] for l in limites]
print(limite_superior)
```

```python
df_pacientes["Ponto Médio"] = [(inferior + superior)/2 for inferior, superior in zip(limite_inferior, limite_superior)]
df_pacientes
```

<p>d) Interpretação da frequência relativa da 3ª classe<p>

```python
total_pacientes = df_pacientes["Nº de pacientes"].sum()
```

```python
df_pacientes["Nº de pacientes"][2]
```

```python
frequencia_relativa_classe = df_pacientes["Nº de pacientes"][2] / total_pacientes *100
int(frequencia_relativa_classe)
```

```python
print(f"{frequencia_relativa_classe}% dos pacientes dormiram entre {df_pacientes["Tempo (h)"][2]} horas.")
```

e) Qual o percentual de pacientes que dormiram menos de 12h?

```python
percent_menos_12h = df_pacientes[df_pacientes["Ponto Médio"] < 12]["Nº de pacientes"].sum() / total_pacientes * 100
percent_menos_12h
```

```python
print(f"Percentual de pacientes que dormiram menos de 12h: {percent_menos_12h:.2f}%")
```
