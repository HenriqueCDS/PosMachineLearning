# Estudo Orientado 1 - Banco de Dados Breast Cancer Wisconsin

Este notebook realiza o download do banco de dados **Breast Cancer Wisconsin** do repositório UCI Machine Learning, 
carrega os dados em um DataFrame Pandas, exibe as primeiras linhas, adiciona os nomes das características 
e realiza o particionamento em conjuntos de treino (80%) e teste (20%).

## 1. Importação das Bibliotecas

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import urllib.request
import os
```

## 2. Download do Banco de Dados

O banco de dados **Breast Cancer Wisconsin (Original)** está disponível no repositório UCI Machine Learning. 
O arquivo `breast-cancer-wisconsin.data` contém 699 instâncias com 11 atributos cada.

```python
# URL do arquivo de dados no repositório UCI
url_data = 'https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/breast-cancer-wisconsin.data'

# Nome do arquivo local
arquivo_local = 'breast-cancer-wisconsin.data'

# Download do arquivo (apenas se ainda não existir localmente)
if not os.path.exists(arquivo_local):
    urllib.request.urlretrieve(url_data, arquivo_local)
    print(f'Download concluído: {arquivo_local}')
else:
    print(f'Arquivo já existe localmente: {arquivo_local}')
```

## 3. Carregamento dos Dados em um DataFrame Pandas

### Descrição das colunas (conforme `breast-cancer-wisconsin.names`):

| # | Atributo | Domínio |
|---|----------|---------|
| 1 | Sample code number (ID) | id number |
| 2 | Clump Thickness | 1 – 10 |
| 3 | Uniformity of Cell Size | 1 – 10 |
| 4 | Uniformity of Cell Shape | 1 – 10 |
| 5 | Marginal Adhesion | 1 – 10 |
| 6 | Single Epithelial Cell Size | 1 – 10 |
| 7 | Bare Nuclei | 1 – 10 |
| 8 | Bland Chromatin | 1 – 10 |
| 9 | Normal Nucleoli | 1 – 10 |
| 10 | Mitoses | 1 – 10 |
| 11 | Class | 2 = benigno, 4 = maligno |

```python
# Nomes das colunas conforme o arquivo breast-cancer-wisconsin.names
nomes_colunas = [
    'id',
    'clump_thickness',
    'uniformity_cell_size',
    'uniformity_cell_shape',
    'marginal_adhesion',
    'single_epithelial_cell_size',
    'bare_nuclei',
    'bland_chromatin',
    'normal_nucleoli',
    'mitoses',
    'class'
]

# Carregamento dos dados em um DataFrame Pandas
# O arquivo usa '?' para representar valores ausentes
df = pd.read_csv(
    arquivo_local,
    names=nomes_colunas,
    na_values='?'
)

print(f'Dados carregados com sucesso!')
print(f'Dimensões do DataFrame: {df.shape[0]} linhas x {df.shape[1]} colunas')
```

## 6. Pré-processamento: Tratamento de Valores Ausentes

A coluna `bare_nuclei` possui 16 valores ausentes representados por `?` no arquivo original. 
As linhas com valores ausentes são removidas para garantir a qualidade dos dados.

```python
df_limpo = df.dropna().reset_index(drop=True)

print(f'Linhas removidas por valores ausentes: {len(df) - len(df_limpo)}')
print(f'Dimensões após limpeza: {df_limpo.shape[0]} linhas x {df_limpo.shape[1]} colunas')
```

## 7. Particionamento dos Dados em Treino (80%) e Teste (20%)

```python

X = df_limpo.drop(columns=['id', 'class'])
y = df_limpo['class']


X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(f'Total de amostras : {len(df_limpo)}')
print(f'Treino (80%)      : {len(X_treino)} amostras')
print(f'Teste  (20%)      : {len(X_teste)} amostras')
```

```python
print(y_treino.value_counts())
print(f'Benigno : {(y_treino == 2).sum()} ({(y_treino == 2).sum()/len(y_treino)*100:.1f}%)')
print(f'Maligno : {(y_treino == 4).sum()} ({(y_treino == 4).sum()/len(y_treino)*100:.1f}%)')

print(y_teste.value_counts())
print(f'Benigno : {(y_teste == 2).sum()} ({(y_teste == 2).sum()/len(y_teste)*100:.1f}%)')
print(f'Maligno : {(y_teste == 4).sum()} ({(y_teste == 4).sum()/len(y_teste)*100:.1f}%)')
```
