```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

data = {
    "Cidade": ["São Paulo", "Rio de Janeiro", "João Pessoa", "Porto Alegre", "Manaus"],
    "Total de Doentes": [19045, np.nan, 8901, 13467, 14569],
    "Total de Curados": [12345, 10890, 5467, np.nan, np.nan],
    "Total de Pessoas que fizeram o tratamento": [np.nan, 13549, 6578, 3467, np.nan]}
df = pd.DataFrame(data)


num_df = df.drop(columns=["Cidade"])

imputer = SimpleImputer(strategy="mean")

num_imputed = imputer.fit_transform(num_df)

df_imputed = pd.DataFrame(num_imputed, columns=num_df.columns)
df_pre_precessado = pd.concat([df["Cidade"], df_imputed], axis=1)

print(df_pre_precessado)

print(f"Médias:{df_pre_precessado.mean(numeric_only=True)}\n")
print(f"Medianas:{df_pre_precessado.median(numeric_only=True)}\n")
print(f"Desvio padrão:{df_pre_precessado.std(numeric_only=True)}\n")
print(f"Amplitude (máx - mín) :{df_pre_precessado.max(numeric_only=True) - df_pre_precessado.min(numeric_only=True)}\n")
```
