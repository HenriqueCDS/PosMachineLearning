```python
df_titanic[['Age', 'Fare']].plot(kind='box', title = " Box-Plot Ideade e tarifas")
```

```python
df_titanic['Age'].plot(kind='hist', bins=30, figsize=(7,4), title='Distribuição das Idades')
plt.xlabel('Idade')
plt.show()
```

```python
df_titanic['Fare'].plot(kind='hist', bins=40, figsize=(7,4), title='Distribuição da Tarifa')
plt.xlabel('Tarifa')
plt.show()
```

```python
df_titanic['Age'].plot(kind='box', figsize=(4,8), title='Boxplot da Idade')
plt.ylabel('Idade')
plt.show()
```

```python
df_titanic['Fare'].plot(kind='box', figsize=(4,6), title='Boxplot da Tarifa')
plt.ylabel('Tarifa')
plt.show()
```

```python
df_titanic.boxplot(column='Fare', by='Pclass', figsize=(7,5))
plt.title('Tarifa por Classe')
plt.suptitle('')
plt.show()
```

```python
df_titanic['Sex'].value_counts().plot(kind='bar', title='Distribuição por Sexo')
plt.show()
```

```python
df_psex = df_titanic.groupby('Sex')['Survived'].mean() * 100

df_psex.plot(kind='bar', title='Taxa de Sobrevivência por Sexo')
plt.ylabel('Taxa de Sobrevivência')
plt.show()
```

```python
df_pclass = df_titanic.groupby('Pclass')['Survived'].mean() * 100

df_pclass.plot(kind='bar')
plt.title('Taxa de Sobrevivência por Classe')
plt.ylabel('Taxa de Sobrevivência (%)')
plt.show()
```

```python
bins = [0, 12, 18, 30, 45, 60, 80]
labels = ['0–12', '13–18', '19–30', '31–45', '46–60', '61–80']

df_titanic['Faixa_Idade'] = pd.cut(df_titanic['Age'], bins=bins, labels=labels)
df_titanic
```

```python


taxa_sobreviv = df_titanic.groupby('Faixa_Idade')['Survived'].mean() * 100

plt.figure(figsize=(8,5))
taxa_sobreviv.plot(kind='bar')

plt.title('Taxa de Sobrevivência por Faixa Etária')
plt.xlabel('Faixa Etária')
plt.xticks(rotation=0)

plt.ylabel('Taxa de Sobrevivência (%)')
plt.grid(axis='y', linestyle='-', alpha=0.5)

plt.show()
```
