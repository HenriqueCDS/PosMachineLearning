# Tratamento e Limpeza de Dados

```python
df_dataframe['Gender'] = df_dataframe['Gender'].str.strip().str.title()
df_dataframe
```

```python
num_cols = df_dataframe.select_dtypes(include=[np.number]).columns
df_dataframe[num_cols] = df_dataframe[num_cols].fillna(df_dataframe[num_cols].median())
df_dataframe
```

```python
text_cols = df_dataframe.select_dtypes(include=['object']).columns
df_dataframe[text_cols] = df_dataframe[text_cols].fillna("Não informado")
df_dataframe
```

```python
for col in num_cols:
    print(col)
    Q1 = df_dataframe[col].quantile(0.25)
    Q3 = df_dataframe[col].quantile(0.75)
    IQR = Q3 - Q1
    limite_inf = Q1 - 1.5 * IQR
    limite_sup = Q3 + 1.5 * IQR
    df_dataframe[col] = np.where(df_dataframe[col] < limite_inf, limite_inf,
                       np.where(df_dataframe[col] > limite_sup, limite_sup, df_dataframe[col]))
df_dataframe
```

 # Engenharia e  Combinação de Dados

```python
df_dataframe["Sleep_Stress_Ratio"] = df_dataframe["Sleep Duration"] / df_dataframe["Stress Level"]
df_dataframe
```

```python
df_dataframe["Age_Group"] = pd.cut(df_dataframe["Age"],
                         bins=[18, 30, 40, 50,65 ],
                         labels=["Jovem", "Adulto", "Meia-idade", "Idoso"])
df_dataframe
```

# Visualizações Univariadas e Bivariadas


Como a distribuição de horas de sono se comporta?

A maioria dorme o recomendado ou há concentração em extremos?



```python
plt.figure(figsize=(8,5))

sns.set_style("darkgrid")  

sns.histplot(
    df_dataframe["Sleep Duration"],
    kde=True,
    color="#4C72B0",
    edgecolor="white",
    linewidth=1.2
)

plt.title("Distribuição da Duração do Sono", fontsize=16, fontweight="bold")
plt.xlabel("Horas de Sono", fontsize=12)
plt.ylabel("Frequência", fontsize=12)

plt.tight_layout()
plt.show()
```

A distribuição de gêneros e por idade no dataset mostra está equilibrada ou existe predominância de algum grupo?



```python
plt.figure(figsize=(7,5))
sns.set_style("darkgrid")
sns.countplot(x="Gender",palette=["#4C72B0","#B04C4C"], data=df_dataframe)

plt.title("Contagem por Gênero")
plt.xlabel("Gênero")
plt.ylabel("Quantidade")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.set_style("darkgrid")
sns.countplot(x="Age_Group",palette='inferno', data=df_dataframe)
plt.title("Contagem por Gênero")
plt.xlabel("Gênero")
plt.ylabel("Quantidade")
plt.show()
```

Existe uma relação perceptível entre quantidade de sono e nível de estresse?

O padrão visual sugere que dormir menos aumenta o estresse?



```python
plt.figure(figsize=(7,5))
sns.set_style("darkgrid")

sns.scatterplot(x="Sleep Duration", y="Stress Level",marker="X", data=df_dataframe)
plt.title("Relação entre Sono e Estresse")
plt.xlabel("Horas de Sono")
plt.ylabel("Nível de Estresse")
plt.show()
```

A relação entre sono e estresse apresenta diferenças visíveis entre os gêneros?



```python
plt.figure(figsize=(7,5))
sns.set_style("darkgrid")
sns.scatterplot(
    x="Sleep Duration",
    y="Stress Level",
    hue="Gender",
     marker="X",
    data=df_dataframe,
   
)
plt.title("Sono vs Estresse por Gênero (Cor como 3ª dimensão)")
plt.xlabel("Horas de Sono")
plt.ylabel("Nível de Estresse")
plt.show()
```

A idade influencia a relação entre horas de sono e estresse, quando observamos o tamanho dos pontos no gráfico multidimensional?

# Visualização de Dados Multidimensionais


Ao comparar as ocupações no FacetGrid, quais profissões parecem apresentar maior nível de estresse relacionado a menor quantidade de sono?

Existem ocupações que mantêm padrões de sono mais regulares do que outras?

Com base na média de horas de sono por ocupação, quais grupos profissionais dormem mais e quais dormem menos?

Os padrões observados nos gráficos sugerem a necessidade de atenção especial para algum perfil (idade, gênero ou ocupação) em relação ao sono e ao estresse?

```python
g = sns.FacetGrid(df_dataframe, col="Occupation", col_wrap=4, height=3.5)
sns.set_style("darkgrid")
g.map_dataframe(sns.scatterplot, x="Sleep Duration", y="Stress Level",color="#3970C9")
g.fig.suptitle("Sono vs Estresse por Ocupação (FacetGrid)", y=1.02)
plt.show()
```

```python
agg_df = df_dataframe.groupby("Occupation")["Sleep Duration"].mean().sort_values()
sns.set_style("darkgrid")
plt.figure(figsize=(10,5))
sns.barplot(x=agg_df.index, y=agg_df.values, palette="cividis")
plt.xticks(rotation=45)
plt.title("Média de Horas de Sono por Ocupação")
plt.xlabel("Ocupação")
plt.ylabel("Média de Horas de Sono")
plt.show()
```
