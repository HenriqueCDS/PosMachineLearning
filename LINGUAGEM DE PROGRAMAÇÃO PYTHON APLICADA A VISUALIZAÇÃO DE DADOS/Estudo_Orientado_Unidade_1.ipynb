```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots()

cidades = ['São Paulo', 'Rio de Janeiro', 'Fortaleza', 'Natal']
quantidade_clientes = [1034, 2345, 678, 456]
bar_labels = ['SP', 'RJ', 'FOR', 'NAT']
bar_colors = ['tab:red', 'tab:blue', 'tab:green', 'tab:orange']


ax.bar(cidades, quantidade_clientes, label=bar_labels, color=bar_colors)
ax.set_title('Quantidade de clientes durante o mês de dezembro por cada filial', fontsize=14, fontweight='bold')
ax.legend(title='Cidades')
plt.show()
```
