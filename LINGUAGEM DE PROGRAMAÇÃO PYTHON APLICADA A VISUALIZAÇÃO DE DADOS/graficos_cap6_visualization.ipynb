```python
import matplotlib.pyplot as plt
import numpy as np
```

```python



capitulos = ["Introdução", "Matplotlib", "Avançado", "Conclusão"]
importancia = [7, 9, 8, 10]

x = np.arange(1, 5)
y = importancia

x_scatter = np.random.rand(20)
y_scatter = 2.5 * x_scatter + np.random.randn(20) * 0.2

y_barras = [15, 30, 25, 20]
dados_hist = np.random.normal(50, 10, 1000)

fig, axs = plt.subplots(2, 2, figsize=(10, 8))
fig.suptitle("Exemplos de Gráficos do Capítulo 6 - Beginning Python Visualization", fontsize=14)

axs[0, 0].plot(x, y, marker="o", color="b", linewidth=2)
axs[0, 0].set_title("Gráfico de Linhas")
axs[0, 0].set_xticks(x)
axs[0, 0].set_xticklabels(capitulos)
axs[0, 0].set_ylabel("Importância")

axs[0, 1].scatter(x_scatter, y_scatter, color="r")
axs[0, 1].set_title("Gráfico de Dispersão (Scatter)")
axs[0, 1].set_xlabel("X")
axs[0, 1].set_ylabel("Y")

axs[1, 0].bar(["Linhas", "Scatter", "Barras", "Histograma"], y_barras, color="g")
axs[1, 0].set_title("Gráfico de Barras")
axs[1, 0].set_ylabel("Frequência")

axs[1, 1].hist(dados_hist, bins=20, color="orange", edgecolor="black")
axs[1, 1].set_title("Histograma")
axs[1, 1].set_xlabel("Valores")
axs[1, 1].set_ylabel("Frequência")

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
