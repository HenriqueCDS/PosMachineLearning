```python
import matplotlib.pyplot as plt

meses = ["abril", "maio", "junho", "julho"]
perdas = [176543, 189021, 198765, 182123]

plt.figure(figsize=(8, 5))
plt.plot(meses, perdas, color='purple', linestyle='-', marker='o',
         linewidth=2, markersize=8, label='Perdas mensais')

# Personalização
plt.title('Perdas de renda ao longo do ano', fontsize=14, fontweight='bold', color='purple')
plt.xlabel('Meses', fontsize=12)
plt.ylabel('Perdas (R$)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(frameon=False, loc='upper left')

plt.ylim(min(perdas) - 4000, max(perdas) + 4000)

plt.show()
```
