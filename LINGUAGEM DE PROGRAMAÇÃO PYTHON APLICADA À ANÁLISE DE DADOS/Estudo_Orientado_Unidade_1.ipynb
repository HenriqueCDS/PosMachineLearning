```python
import requests
from bs4 import BeautifulSoup
```

```python
cidades = ["Campinas", "São Paulo", "Curitiba", "Recife"]
```

```python
base_url = "https://cidades.ibge.gov.br/brasil/sp/"
```

```python
estados = {
    "Campinas": "sp",
    "São Paulo": "sp",
    "Curitiba": "pr",
    "Recife": "pe"
}
```

```python
def obter_densidade(cidade):
    estado = estados[cidade]
    url = f"https://www.ibge.gov.br/cidades-e-estados/{estado}/{cidade.lower().replace(' ', '-')}"
    print(url)
    response = requests.get(url)
    if response.status_code != 200:
        print(f"Erro ao acessar {cidade}: {response.status_code}")
        return None
    
    soup = BeautifulSoup(response.text, "html.parser")
    
    elemento = soup.find(string="Densidade demográfica")

    if elemento:
        print(elemento)
        valor = elemento.find_next("p").text.strip()
        return valor
    else:
        print(f"Densidade demográfica não encontrada para {cidade}.")
        return None
```

```python
densidade_por_cidade = []
for cidade in cidades:
    densidade = obter_densidade(cidade)
    numero_densidade = float(densidade.split()[0].replace(".", "").replace(",", "."))

    if densidade:
        densidade_por_cidade.append({
            cidade:numero_densidade  
        })
        print(f"{cidade}: {densidade}")
```

```python
densidade_por_cidade
```

```python
densidade_por_cidade_dict = {cidade: densidade for d in densidade_por_cidade for cidade, densidade in d.items()}
densidade_por_cidade_dict
```

```python
for cidade, densidade  in densidade_por_cidade_dict.items():
  
    print(f"\nCidade: {cidade}")
    print(f"Densidade demográfica: {densidade:.2f} hab/km²")

    if densidade > 110:
        print("O parque PODE ser implantado nesta cidade.")
    else:
        print("O parque NÃO pode ser implantado nesta cidade.")
```
