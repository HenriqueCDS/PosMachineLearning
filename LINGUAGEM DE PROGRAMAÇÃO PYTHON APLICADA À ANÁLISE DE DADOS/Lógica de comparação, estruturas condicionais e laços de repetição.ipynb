```python
## exercício 1 - AULA 4
def fatorial(numero):
    while True:
        if numero.isdigit():
            fatorial = 1
            for i in range(1,int(numero)+1):
                fatorial *=  i  
            return fatorial
entrada = input("Digite um numero")
print(f"Fatorial: {fatorial(entrada)}")
```

```python
## exercício 3 - AULA 4
numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input("Digite o segundo número: "))

def maximo_divisor_em_comum(numero1, numero2):
    if numero1 > numero2:
        maior_num = numero1
        menor_num = numero2
    else:
        maior_num = numero2
        menor_num = numero1

    while menor_num != 0:
        resto = maior_num % menor_num
        maior_num = menor_num
        menor_num = resto

    return maior_num

resultado = maximo_divisor_em_comum(numero1, numero2)
print("O Maximo divisor em comum é:", resultado)
```

```python
## exercício 4 - AULA 4

for i in range(100,501):
    num_str = str(i)
    num_digitos = len(num_str)
    soma = 0
    for digito in num_str:
        soma += int(digito) ** num_digitos
    if soma == i:
        print(i, "é um número de Armstrong.")
 
```

```python
## exercício 1 - AULA 5
texto = input("Digite um texto").upper()
qtd_texto,qtd_numeric = 0,0
for i in range(len(texto)):
    if texto[i].isalpha():
        qtd_texto += 1 
    elif texto[i].isdigit():
        qtd_numeric += 1
print(f"Quantidade de letras encontradas:{qtd_texto}")
print(f"Quantidade de numeros encontrados:{qtd_numeric}")
```

```python
## exercício 2 - AULA 5
def calculadora_operacao(num,num2,operador ):
    if operador  == 1:
        print("------------------------------------")
        print("ADIÇÃO")
        print(f"{num} + {num2}")
        print("------------------------------------")
        return num+num2
    elif operador  == 2:
        print("------------------------------------")
        print("SUBTRAÇÃO")
        print(f"{num} - {num2}")
        print("------------------------------------")
        return num-num2
    elif operador  == 3:
        print("------------------------------------")
        print("MULTIPLICAÇÂO")
        print(f"{num} * {num2}")
        print("------------------------------------")
        return num*num2 
    elif operador  == 4:
        print("------------------------------------")
        print("DIVISÃO")
        print(f"{num} / {num2}")
        print("------------------------------------")
        return num/num2
    else: 
        print("Nenhum operador correspodende foi encontrado")

while True:
    num = input("Digite um numero:")
    num2 = input("Digite um numero 2:")
    operador = input("Digite um numero correspondente a um opção da lista: 1 - adição, 2 - subtração, 3 - multiplicação, 4 - divisao")

    if(num.isdigit() and num2.isdigit() and operador.isdigit()):
        num,num2,operador = int(num),int(num2),int(operador)
        resultado = calculadora_operacao(num,num2,operador)
        print(f"Resultado da operação foi de {resultado}")
        print("------------------------------------")
        break
    else:
        print("Erro: Digite apenas numeros!!")
        print("------------------------------------")
```

```python
## exercício 3 - AULA 5
def fibonacci(n):
    aux, aux2 = 0, 1
    for _ in range(n):
        print(aux, end=" ")
        aux, aux2 = aux2, aux + aux2
while True:
    numero = input("Quantos numeros da sequencias serão imprimidos")
    if numero.isdigit():
        print("Sequencia de fibonaci:")
        fibonacci(int(numero))
        break
    else:
        print("ERRO: digite ua numero")     
```
