```python
## exercício 3 - AULA 2
def calcular_salario_liquido(salario_bruto):
    imposto_renda = salario_bruto * 0.15
    print(f"Desconto do imposto de renda: {imposto_renda}")
    inss = salario_bruto * 0.11
    print(f"Desconto do INSS: {inss}")
    sindicato=  salario_bruto * 0.05
    print(f"Desconto do Sindicato: {sindicato}")
    total_descontos = imposto_renda+inss+sindicato
    print(f"Total de Descontos: {total_descontos}")
    salario_liquido = salario_bruto - total_descontos
    print(f"Seu salario líquido  é {salario_liquido}")
calcular_salario_liquido(float(input("Qual é o seu salario bruto")))    

```

```python
## exercício 1 - AULA 3
def bonus_quinquenio(salario,ano_sevico):
    if ano_sevico > 5:
        return salario * 0.05
    return 0

salario = float(input("Me informe seu salario"))
anos_sevico =int(input("Me informe seu anos de serviço"))

bonus_liquido = bonus_quinquenio(salario,anos_sevico)

if bonus_liquido > 0:
    print(f"Seu bonus liquido de quinquenio {bonus_liquido}")
else: 
    print(f"Seu tempo de anos de seviço ainda não atigiram o tempo sufiente para solicitar o bonus")


```

```python
## exercício 2 - AULA 3
def verificar_suspeitos():
    classificar_assasino = []
    perguntas = ["Telefonou para a vitima?",
                "Esteve no local do crime?",
                "Mora perto da vitima?", 
                "Devia para a vitma?",
                "Já trabalhou com a vitima?"]
    for i in range(0,5):
        while True:
            resposta = input(f"{perguntas[i]} - digite 1 se sim ou 0 se não")
            if resposta in ("0", "1"):
                classificar_assasino.append(int(resposta))
                break
            else:
                print("Digite um valor correto")
    valor_repostas = sum(classificar_assasino)
    print(valor_repostas)
    if valor_repostas == 2:
        return "suspeito"
    elif valor_repostas >= 3 and valor_repostas < 5: 
        return "Cúmplince"
    elif valor_repostas == 5:
        return "Assassino" 
    else:
        return "Inocente"

classificado = verificar_suspeitos()
print(f"Voçê está classificado como {classificado}")

```

```python
## exercício 3 - AULA 3

print("==============================================================================")

print("Carrinho de compras")

print("==============================================================================")

def carrinho_de_compras():
    preco_total = 0

    while True:
        nome_produto = input("Nome produto:")
        preco_produto = input("Preço produto:")
        qtd_produtos = input("Quantidade de produtos:")
        if nome_produto.isdigit() == False  and qtd_produtos.isdigit() and preco_produto.isdigit():
                Fechamento_carrinho = input("Encerrar carrinho de compras ( se sim digite 0 ou se não digite qualquer outro valor ):")
                qtd_produtos = int(qtd_produtos)
                preco_produto = float(preco_produto)
                
                print("-------------------------------------------------------------------------------")
                print("Nome Produto:",nome_produto)
                print("Preço do produto: ", preco_produto)
                print("Quantidade de produto: ", qtd_produtos)
                preco_produto = qtd_produtos * preco_produto
                print("Preço total do produto: ",preco_produto)
                print("-------------------------------------------------------------------------------")
    
                preco_total += preco_produto
                if Fechamento_carrinho == '0':
                    return preco_total  

        else:
            print("Erro: digite um número inteiro válido.")


preco_total = carrinho_de_compras()

Filiado = input(f"Você é filiado ao programa de ponto da loja para garantir seu desconto de 10% em sua compra - digite 1 se sim ou 0 se não")  

print("==============================================================================")
if Filiado == "1":
            desconto_filiado = preco_total * 0.1
            print(f"Desconto de filiado: {desconto_filiado}")
            preco_total = preco_total - desconto_filiado 
            print(f"Preço total de sua compra será {preco_total}")
elif Filiado == "0":
            print(f"Preço total de sua compra será {preco_total}")
print("==============================================================================")

    
```

```python
## exercício 4 - AULA 3
def classificador_idade(idade):
    if idade == 11:
        return "Criança"
    elif idade >= 12 and idade <= 17:
        return "Adolecente"
    elif idade >= 18 and idade <= 35:
        return "Jovem"
    elif idade >= 36 and idade <= 59:
        return "Adulto"
    else:
        return "Idoso"

print("-------------------------------------------")
print("Classificação etária")
print("-------------------------------------------")

while True:
    idade_usuario = input("Qual sua idade?")
    if idade_usuario.isdigit():
        idade_usuario = int(idade_usuario)
        print(f"Voce esta classificado como {classificador_idade(idade_usuario)}")
        break
    print("Digite um idade valida")

```
